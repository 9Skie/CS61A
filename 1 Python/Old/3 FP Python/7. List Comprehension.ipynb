# List Comprehension

Just a better way to work with lists, making em, alterating em, and everything with em in a easier, stylish mannor

Say we got a list, and we want to create a new list with all the values squared

But uh... didn't we already learn how to do that with map?

Yes, but list comprehension is also a valid choice


Let's first see it in map function style

```python
nums = [1, 3, 5, 9, 3, 4, 6, 8, 14, 31]

square_nums = list(map(lambda num: num*num, nums))
print(square_nums)
```

Now, let's see it in list comprehension style

```python
nums = [1, 3, 5, 9, 3, 4, 6, 8, 14, 31]

square_nums = [num*num for num in nums]
print(square_nums)
```

And it's not just that! Let's look at another example, where I want a list of numbers from the list nums but only the even numbers.

We learnt how to do this with the filter function, right? Let's see that first

```python
nums = [1, 3, 5, 9, 3, 4, 6, 8, 14, 31]

even_nums = list(filter(lambda num: num % 2 == 0, nums))
print(even_nums)
```

But, list comprehention can also do that, just add an if statement at the end of the function

```python
nums = [1, 3, 5, 9, 3, 4, 6, 8, 14, 31]

even_nums = [num for num in nums if num % 2 == 0]
print(even_nums)
```

Being honest, list comprehention looks like a mixture of lambda functions with for loops

let's write out the equvilant statement of sorting even numbers in a for loop

so we understand the syntax of list comprehension better

```python
nums = [1, 3, 5, 9, 3, 4, 6, 8, 14, 31]
even_nums = []

# this has the same functionality as "even_nums = [num for num in nums if num % 2 == 0]"
for num in nums:
    if num % 2 == 0:
        even_nums.append(num)

print(even_nums)
```

Now looking at the list comprehension syntax:

- the returned value is at the front
- the for loop is in the middle
- and the conditional is at the back

And, while it is called list comprehension, we can do this for other iterable data types, like dicts, sets

Here is a list comphresion of creating a dict of key values pairs of super hero names, to their actual names

```python
names = ['Bruce', 'Clark', 'Peter', 'Logan', 'Wade']
heros = ['Batman', 'Superman', 'Spiderman', 'Wolverine', 'Deadpool']

hero_and_actual_names = {name: hero_name for name, hero_name in zip(names, heros)}
print(hero_and_actual_names)
```

Here's another list comprehension of creating a set from a list of values

```python
nums = [1,1,2,1,3,4,3,4,5,5,6,7,8,7,9,9]

nums_set = {num for num in nums}
print(nums_set)
```

But... you can have a for in a for!
