# Lambda Functions

Remember from our previous higher order functions?

Most of them had to take in a function as a way to manipulate the data from the iterable object

Though, most of the times these functions are just used 1 off for that specific purpose 

and never to be seen again any where else in the code

But we still have to do all this formal thing, like naming it, telling it wall parameters it takes, doing this operation, then returning something... rather tedious

Lambda functions aim to solve that, they are 1 off functions that are created for that specific use, then never to be seen again.

They are called annonymous functions, as there isn't a way to call them, they are not bound to some identifier

But again, let's begin with an example, where we are just cubing all the values of a list, and returning the list

This would be how we had done it, using our recently learnt map function

```python
def cube(number):
    return number * number * number

nums = [3, 5, 9, 12]
cube_nums = list(map(cube, nums))
print(cube_nums)
```

But we can do this instead with a lambda function

```python
nums = [3, 5, 9, 12]
cube_nums = list(map(lambda num: num*num*num, nums))
print(cube_nums)
```

The syntax does look a little weird at first... let me explain with another example

A lambda function is divided into 3 parts, the lambda word, the function's parameters, and the operation

You can assign it to a variable... but doing that basically has no difference compared to just defining a function then calling it

```python
add_2 = lambda number: number + 2

print((lambda number: number + 2)(5))
print(add_2(5))
```

And that was it for lambda functions! Really simple stuff that looks hard on the surface
