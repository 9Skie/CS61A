# *Args and **Kwargs

Let's first explain what those * are doing around here, it does an operation called unpacking

```python
nums = [1,3,5,9,7]

print(nums)
print(*nums)
```

If the * is placed by an iterable, it means that it expands the iterable into it's individual elements

but that does not mean to use however many variables to capture individual elements

```python
# this is not allowed python syntax, there's an error
num1, num2, num3, num4, num5 = *nums
```

But unpacking doesn't equal to Args Kwargs, in fact, it's almost alike a reverse.

Args would take in these individual elements passed to the function, and group them together into a tuple

We can call this composed tuple from passed parameters through *args


let me give an example of where this can be helpful.

Say there is a function that is called "order pizza", and by default the order takes in the size of the pizza.

But there's a problem, toppings can vary from pizza to pizza, and your ordering system needs to take an arbitrary number of potential toppings the customers might want ot add, and so args can save the day

```python
def order_pizza(size, *args):
    print(f"There's an order of size {size} pizza")
    print()
    print("Here are the pizza's toppings:")
    for arg in args:
        print(f"- {arg}")


order_pizza("Large", "Beef", "Potatoes", "Cheese", "Onions", "Ketchup")
```

But... while conventionally we call it args, we can actually name it however we like, toppings make much more sense in this case

```python
def order_pizza(size, *toppings):
    print(f"There's an order of size {size} pizza")
    print()
    print("Here are the pizza's toppings:")
    for topping in toppings:
        print(f"- {topping}")


order_pizza("Large", "Beef", "Potatoes", "Cheese", "Onions", "Ketchup")
```

Ok... but what about the **kwargs? If order of how we passed the parameters to the function matters, and *args captures everything behind the basic parameters, how does **kwargs know what values to operate on?

Well... *args makes a set from all the positional arguments after the essential ones, **kwargs makes a dictionary from later passed key value pairs to the function

```python
def order_pizza(size, *args, **kwargs):
    print(f"There's an order of size {size} pizza")
    print()
    print("Here are the pizza's toppings:")
    for arg in args:
        print(f"- {arg}")
    print()
    print("Here are the details of the order:")
    for key, value in kwargs.items():
        print(f"- {key} : {value}")

order_pizza("Large", "Beef", "Potatoes", "Cheese", "Onions", "Ketchup", delivery=True, delivery_driver="Johnny Hopkin", tip=5, est_arrive_time="13:00pm")
```

Course... you can do the custom naming thing as well for kwargs

```python
def order_pizza(size, *toppings, **order_details):
    print(f"There's an order of size {size} pizza")
    print()
    print("Here are the pizza's toppings:")
    for topping in toppings:
        print(f"- {topping}")
    print()
    print("Here are the details of the order:")
    for key, value in order_details.items():
        print(f"- {key} : {value}")

order_pizza("Large", "Beef", "Potatoes", "Cheese", "Onions", "Ketchup", delivery=True, delivery_driver="Johnny Hopkin", tip=5, est_arrive_time="13:00pm")
```
