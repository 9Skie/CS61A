# Higher Order Functions

Now, we just understood about how to iterate through our objects with iterators, and a "memory friendly" iterator called the generator.

The next thing we are learning are functions that manipulate the returned individual elements from the iterators through some function.

Now, why would we do this with new functions to learn? Isn't a for loop enough to iterate through the data, and manipulate individual elements for all this?

True... these functions do the same thing, but just in a more concise way, who doesn't love elegant code?

The true name of these functions are called higher order functions, as they are functions that take in functions

(and maybe, you can even design them yourself to return functions!)

# Map

Say I wanted to iterate through a list of numbers, square each individual element and create a new list from those values.

This would be how we'd do it through a for loop

```python
def square(number):
    return number * number

nums = [2, 4, 6, 8, 10]
square_nums = []

for num in nums:
    square_nums.append(square(num))

print(square_nums)
```

But with the map function, we can take in a function, and an iterable

And map would return an iterator apply that function to every individual element of the iterable

But since it returns an iterator... we just apply the list class to the iterator

```python
def square(number):
    return number * number

nums = [2, 4, 6, 8, 10]

square_nums =  list(map(square, nums))
print(square_nums)
```

And as the name suggests, the map function "maps" values in an iterable to another value through the given function

# Filter

Say I wanted to iterate through a dict of houses, but only keep the houses where the price per unit square is less than $500

This would be how we do it through a for loop

```python
avaliable_houses = {
    "615 N Maple St, North Platte, NE 69101" : 1000,
    "710 N Country Club Dr, Colby, KS 67701" : 250,
    "1002 Union Ave, North Platte, NE 69101" : 700,
    "520 Coke St, Bartley, NE 69020" : 350,
    "704 E 1st St, Trenton, NE 69044" : 500,
}

def price_comparison(house_price):
    return True if house_price < 500 else False

desireble_houses = {}

for house, house_price in avaliable_houses.items():
    if price_comparison(house_price) == True:
        desireble_houses.update({house : house_price})

print(desireble_houses)
```

And since we saw what the function map does, filter basically does the same

but as the name suggests, it "filters" out values in an iterable if it doesn't meet the criteria of the function 

```python
avaliable_houses = {
    "615 N Maple St, North Platte, NE 69101" : 1000,
    "710 N Country Club Dr, Colby, KS 67701" : 250,
    "1002 Union Ave, North Platte, NE 69101" : 700,
    "520 Coke St, Bartley, NE 69020" : 350,
    "704 E 1st St, Trenton, NE 69044" : 500,
}


def price_comparison(house_name):
    house_price = avaliable_houses[house_name]
    return True if house_price < 500 else False

desireble_houses = list(filter(price_comparison, avaliable_houses))
print(desireble_houses)
```

# Zip

say I had a set of student names, and a set of student grades, and I wanted to combine them so each student has a corresponding grade

This would be how we'd do it through a for loop

```python
grades = [37, 93, 78, 63]
names = ["John", "James", "Julia", "Jones"]
students = []

if len(grades) == len(names):
    for index in range(len(names)):
        name = names[index]
        grade = grades[index]
        students.append((name, grade))

print(students)
```

But with the Zip function, we can take elements from two or more iterables

and create a zip object with paired elements stored in tuples for each element

And of course, we need to turn it into a list for our specific use case

```python
grades = [37, 93, 78, 63]
names = ["John", "James", "Julia", "Jones"]

students = list(zip(names, grades))
print(students)
```

# Sort

You know what? Let's keep using students as an example, say I wanted to sort the students by their grades descending

This would be how we would do it through a bubble sort

```python
students = [('John', 37), ('James', 93), ('Julia', 78), ('Jones', 63)]


is_sorted = False
stu_len = len(students)

while not is_sorted:
    is_sorted = True
    for index in range(stu_len - 1):
        if students[index][1] < students[index + 1][1]:
            temp = students[index]
            students[index] = students[index + 1]
            students[index + 1] = temp
            is_sorted = True

print(students)
```

Tedious to do, isn't it? Thankfully python has the sort function implemented for all it's data types

and you can even call it to sort under a ceriteria of a function, in this case we sort the list by the student's grades

```python
def get_grade(student):
    return student[1] # basically returns the grades of the student

students = [('John', 37), ('James', 93), ('Julia', 78), ('Jones', 63)]
students.sort(key=get_grade, reverse=True) # a list is not returned! the list is being sorted in place
print(students)
```

But... this sort function is specifically dedicated to lists, and not other iterables, so let's try this again with a more general sort function

And instead of sorting by their grades, let's sort by their names ascending, and turn it into a tuple first before sorting lol

```python
def get_name(student):
    return student[0] # basically returns the grades of the student

students = (('John', 37), ('James', 93), ('Julia', 78), ('Jones', 63))
sorted_students = sorted(students, key=get_name, reverse=False) # in this case a new (whatever iterable data type it is) is created and returned, depending on the input data type
print(sorted_students)
```
