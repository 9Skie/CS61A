# Recursion

Recursion is a programming technique where a function calls itself to solve smaller instances of the same problem.

This process continues until a specified base condition is met. 

Once the base condition is fulfilled, the function stops calling itself and starts to return values

and combines the results from all the recursive calls in a defined manner.

Let's take a very simple example of getting a number to it's nth power, 

this would be how we do it with a simple function with a for loop

```python
def nth_power(number, n):
    result = number
    for _ in range(n-1):
        result *= number
    return result

print(nth_power(2, 3))
```

And this would be how we'd do it in recursion!

```python
def nth_power_recursively(number, n):
    if n <= 1:
        return number
    else:
        return number * nth_power_recursively(number, n - 1)
    

print(nth_power_recursively(2, 3))
```

Honestly, recursion is simply just a conditional loop like any other loop.

Let's break down what is actually going on behind the scenes for this recursion to happen using our example where

- number is 2
- n is 3

number * nth_power_recursively(number, n - 1) would turn into...

2 * nth_power_recursively(number, 3 - 1)

= 2 * (2 * nth_power_recursively(number, 2 - 1))

= 2 * (2 * 2) 

= 2 * 4

= 8

Because... we set a conditional where if n reached 1 or less, the function would stop recursion annd not return itself

And that is basically it for recursion! Sure you can make more commplex recursive functions, but there's just 2 main underlying points

- A way to combine the recursive calls together (in this case, multiplying them together)

- A way to step out of the recursive calls (in this case, when n <= 1)

This is another example, but a factorial

```python
def factorial(num):
    if num == 1:
        return 1
    return num  * factorial(num - 1)
```

```python
print(factorial(5))
```
