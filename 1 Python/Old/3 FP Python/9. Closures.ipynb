# Closures

A closure is an inner function that remembers and has access to variables in the local scope 

in which it was created even after the outer function has finished executing

Uh... didn't we just look at an example of that in first class functions?

Yeah, so let's do another little funny thing here

```python
def fancy_line(kamoji_1, kamoji_2):

    def print_fancy_line(text):
        print(f"{kamoji_1} | {text} | {kamoji_2}")

    return print_fancy_line


love_line = fancy_line("(*♥ω♥)", "(♥ω♥*)")
pretty_line = fancy_line("(✿◠‿◠)", "(◠‿◠✿)")
wink_line = fancy_line("(｡•̀ᴗ-)✧", "✧(-ᴗ•̀｡)")

love_line("THIS")
pretty_line("IS")
wink_line("FANCY TEXT!")
```
