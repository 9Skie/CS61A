# Type Hinting

Type hinting in Python just is a way of specifying the expected data types of variables, function parameters, and return values in the code.

It's a simple concept that's actually FORCED in most other programming languages, just not python (and I fucking love that in most cases)

But... it can be a little confusing when we read documentation on imported functions from libraries, since we don't understand what they do.

Normally, we'd write a function somewhat like this in python ways

```python
def add(x, y):
    return x + y

print(add(10, 3))
```

This works well enough for simple functions, because it's pretty intuitive what you'd expect it to take in, and what it spits out.

But for more complex functions with more naming conventions, it gets tedious to read and understand, so simply... we add annotations, and type hinting

We can state the datatype of input variables, what datatype will the function return

```python
def add(x: int, y: int) -> int:

    """
    This functions does addition between two integers
    """
    
    return x + y
```

The output wouldn't change... it just gets easier to read by developers, and you from the future when you look back at your mess of a code

```python
print(add(10, 3))
```
