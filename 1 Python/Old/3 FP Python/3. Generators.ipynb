# Generators

A very close friend of iterators! But with a slight difference...

Say I have a text file, and I wanted to get each line from this file

```python
def fetch_lines(filename):
    with open(filename, "r") as txt_file:
        lines = []
        for line in txt_file:
            lines.append(line)
    return lines


zen_poem = fetch_lines("example.txt")
for line in zen_poem:
    print(line)
```

In our current case this is fine, the entire text file get's stored in memory in a single list.

But... what if we have huge files that we want to iterate through? Maybe not just files, just some sort of data, larger than the amount of avaliable memory.

If we try that, a Memory error occurs.

Then that's where generators come into play.

Instead of getting all the data and stored in memory at once, why don't we just load 1 piece of data at a time?

Do whatever we need to do with it, then just erase it from memory and move onto the next piece of data?

Generators do exactly that, load data at a time into memory.

```python
def fetch_lines(filename):
    with open(filename, "r") as txt_file:
        for line in txt_file:
            yield line

poem = fetch_lines("test.txt")
print(poem)
```

Ok! By changing our function to "yield" the line instead of returning all the lines, we get a generator object

This actually dips into something else called a context manager for file handling... we aren't gonna go deep about those

But how do we read a line from a generator? Same as a iterator! With `next()`

```python
def fetch_lines(filename):
    with open(filename, "r") as txt_file:
        for line in txt_file:
            yield line

poem = fetch_lines("example.txt")
print(next(poem))
print(next(poem))
print(next(poem))
print(next(poem))
```

As you might have guessed from our iterators video, you can use a generator (yield keyword) to easily create custom iterables to use in for loops

This can also work for all the iterables that we've talked about before, but... yield is not a keyword allowed outside a function, because we have to assign it to a variable to capture the generator

```python
def get_items(a_list):
    for item in a_list:
        yield item

my_list = [1, 2, 3, 4, 5]
my_list_generator = get_items(my_list)
print(next(my_list_generator))
print(next(my_list_generator))
print(next(my_list_generator))
print(next(my_list_generator))
print(next(my_list_generator))
```
