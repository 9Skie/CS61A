# Iterables

 An iterable is an object that is capable to returning it's elements 1 at a time.

 This includes lists, tuples, dicts, sets, strings, even files

```python
my_list = [1, 2, 3, 4, 5]
for item in my_list:
    print(item)
```

```python
my_tuple = (1, 2, 3, 4, 5)
for item in my_tuple:
    print(item)
```

```python
my_dict = {1:"a", 2:"b", 3:"c", 4:"d", 5:"f"}
for key in my_dict:
    print(key)
```

```python
my_set = {1, 2, 3, 4, 5}
for item in my_set:
    print(item)
```

```python
my_string = "hello"
for char in my_string:
    print(char)
```

```python
file_path = "example.txt"

with open(file_path, "r") as txt_file:
    for line in txt_file:
        print(line)
```

But... what's going in the background for these things to be "iterable"?

It actually quite nicely ties into our OOP series, since these are special methods for classes to define "what it means to iterate through them", called `__iter__`

When you iterate through an iterable in Python, it returns an iterator object, which defines what it means to iterate through the elements of the iterable one at a time

These data types that we've just shown examples of already have a definition for what it means to iterate through them.

But... we can also create our custom classes, and define what it means to iterate through them as well


Now, what's an iterator? While we said it's a definition of what it means to return 1 element at a time of an object, there's a more concise definition.

An iterator remembers its own internal state, basically where it is during the iteration, and what is the next value in the next iteration.

Iterable Objects are not Iterators, they are separate things, when we try to iterate through an iterable object, it returns an iterator.

(Man this sounds like some sort of tongue twister...)

So let's see an example, we can directly get the iterator for an iterable object

```python
nums = [1, 2, 3, 4, 5]

# item_nums = iter(nums) also works
iter_nums = nums.__iter__()

print(iter_nums)
```

But how does an iterator iterate? Well, that has to do with the `__next__` method

Also, since iterators remember where it is during iteration, so if we run `__next__` again and again, it remembers where it left off from the last  `__next__` and iterates to the next element

```python
print(next(iter_nums))
print(next(iter_nums))
print(next(iter_nums))
print(next(iter_nums))
print(next(iter_nums))
```

But... if we keep going after we've iterated through every element of the object, we get an error

basically saying there are no more elements to iterate, you should stop

Though if this was just a for loop, it automatically knows when to end the iteration

```python
print(next(iter_nums))
```

So, this is some what the true code going underneath a for loop in python

```python
iter_nums = iter(nums)

while True:
    try:
        item = next(iter_nums)
        print(item)
    except StopIteration:
        break
```

# Range

You know the range() function that we always use in python to iterate through a certain number of values?

Let's try to make the code that underlies that function in a simplistic sense, so where we can say a start and end value

(Because you can pass in 1 number to say 0 - that number, 2 numbers to say start number - end number, 3 numbers to say start number - end number, and the steps to take)

```python
class My_Range:

    def __init__(self, start, end):
        self.value = start
        self.end = end

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.value >= self.end:
            raise StopIteration
        current_value = self.value
        self.value += 1
        return current_value
```

Now, an iterator always returns itself when the `__iter__` method is called upon, so we just return itself

As for the `__next__` method, we need to think about 2 things

- How do we normally just move to the next value in iteration?
- How do we know when to stop? Basically when have we iterated to the end of the range?

normally iterating to the next value is easy, we simply increment self.value by 1

and when does it end? when self.start >= self.end, because the range function in python only iterates up to the end value, but not including it

```python
nums = My_Range(1, 10)

for num in nums:
    print(num)
```

# Words of a Sentence

Say we were given a sentence like "Today is a sunny day" we need our iterator to break it down into individual words and iterate through like

- Today
- is
- a
- sunny
- day

How should we create an iterator to do this? (Assume there are no punctuation! Then it would be a very difficult question)

```python
class Sentence:
    
    def __init__(self, sentence):
        self.words = sentence.split()
        self.word_count = len(self.words) - 1
        self.index = 0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index > self.word_count:
            raise StopIteration
        
        current_word = self.words[self.index]
        self.index += 1
        return current_word
```

```python
a_sentence = Sentence("What a sunny day to surf on the waves")

for word in a_sentence:
    print(word)
```
