# First Class Functions

We just learnt about higher order functions, which works on other functions, meaning that they take one or more functions as an argument and can also return a function.

So what the heck are these? 

Well, you know how we normally call a function, we usually go `function_name()`, there's a parenthese to indicate that this is a function, alongside with whatever it takes in

For first class functions, it means that we can assign `function_name()` to a variable, and pass it around

When you pass a function `function_name` without parentheses, you are passing the function object itself. This allows the function to be called or used as an argument within another function.

When you pass a function `function_name()`  with parentheses, you are calling the function and passing its return value.

```python
def square(num):
    return num * num

print(square)
print(square(5))
```

But also... you can just assign that function to a variable, and it would act exactly like the function

```python
variable = square

print(variable)
print(variable(5))
```

Say we wanted to make a function that says hi to people, but different people would recieve different greetings based on their name.

Also, there might be different greetings in different scenarios, like say different languages

This would be how we do it with a conditional function

```python
def greeting(prefix, name):
    return f"{prefix}, {name}!"

print(greeting("Bonjour", "Alice"))
print(greeting("Ciao", "Bob"))
```

This is how we can do it with a first class function (which is defined by a higher order function)

```python
def greeting_v2(prefix):
    def greet(name):
        return f"{prefix}, {name}!"
    return greet


# These are first class functions
greet_french = greeting_v2("Bonjour")
greet_italian = greeting_v2("Ciao")

print(greet_french("Alice"))
print(greet_italian("Bob"))
```

Now, what's the difference here?

In our simple function approach, every time when we call the function, we have to specify the prefix of which language it is.

Where by using a first class function, we first can save the functioni in a variable for a respective language.

Then we can directly call the function with a name to get a greeting in the respective language.

There's no need to pass in the prefix anymore explicitly every time, the returned function itself remembers the prefix.

You can already tell for yourself that our code would become more concise in the long run, and this ties into our next topic as well, closures.
