# Dunder Methods

Dunder methods, short for "double underscore methods" are methods in python that python would recognize for how to perform specific operations on an object

They are also called Magic Methods... I really wonder who came up with these names in the first place

Let's take a very simple example, adding 2 integers and adding 2 strings

```python
print(1 + 2)
print("Hi" + " John")
```

When we "add" an object to an object, (yes, integers and strings are objects) Python looks for the `__add__()` method in the objects that it's doing addition on.

Like in this case, "add" for integers just mean getting the sum of the two integers.

While "add" for strings mean something else, concatenating the contents of both strings together.

This code below will perform the exact same functionality as above, it's just how python truly does it in the underlying code

```python
print(int.__add__(1, 2))
print(str.__add__("Hi", " John"))
```

There are tons of these Dunder methods that we can define for our class, Employees, if you think about it

What would it mean to:
- Add two employees together?
- Print out the contents of an employee?
- For an employee to equal another employee?
- Any other wild ideas for what it makes sense for employees to do? 

```python
class Employee:

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return '{} {}'.format(self.first, self.last)
```

Let's start by printing out the contents of an employee

```python
emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)

print(emp_1)
```

That... really just gives where the object is in memory, and is the default implementation of the Dunder method `__repr__()` and `__str__()`, both make a string, but serve different purposes

- `__repr__()` is more used to provide information that can be used to recreate the object
- `__str__()` is more used to be readable and informative to users of what the object is

But that is not informative for seeing what an employee is made of, so we need to implement it ourselves

```python
class Employee:

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"

    # this is for providing how to create this object
    def __repr__(self):
        pass

    # yay! our new definition for what it means to print the employee out as a string
    def __str__(self):
        return f"{self.fullname()} - {self.email}"
```

Now let's see what printing out our employee gets us

```python
emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)

print(emp_1)
```

That works fine now, what about adding two emplyees?

I can see the most intuitive meaning of that being trying to add their salary together, maybe seeing how much money the company pays in total for all employees? Who knows

```python
class Employee:

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"

    def __repr__(self):
        pass

    def __str__(self):
        return f"{self.fullname()} - {self.email}"
    
    # how to add two employees!
    def __add__(self, another_employee):
        return self.pay + another_employee.pay
```

So, this should return their sum salary now

```python
emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)

print("Total salary is: " + str(emp_1 + emp_2))
```

Or the length of the employee's name?

```python
class Employee:

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"

    def __repr__(self):
        pass

    def __str__(self):
        return f"{self.fullname()} - {self.email}"
    
    def __add__(self, another_employee):
        return self.pay + another_employee.pay
    
    def __len__(self):
        return len(self.fullname())
```

Let's see another example...

```python
emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)

print("Length of name is: " + str(len(emp_1)))
```
