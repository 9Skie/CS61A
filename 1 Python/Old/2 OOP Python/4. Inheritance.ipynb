# Inheritance

So, what is the big idea of this?

Inheritance allows us to make classes, which can directly use the methods and attributes of another class.

It means we can reuse code, overwrite the methods, and add additional pieces of functionality if needed

In this case, we have an Employee Class, but there can be many types of Employees that have their own distinctions.

Like Developers, Accountants, Human Resources, Marketing, Executives, Interviewers... Tons of different types of people, that all are employees, but differ a bit in some way.

```python
import random

class Employee:

    break_allowance_days = 4

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return '{} {}'.format(self.first, self.last)
    
    def eligible_for_break(self):
        if self.days_worked > Employee.break_allowance_days:
            return True
        else:
            return False
            
    @classmethod
    def define_allowance_days(cls, days):
        if days < 7:
            cls.break_allowance_days = days 

    @staticmethod
    def generate_ID():
        ID = random.randint(10000000, 99999999)
        return "Employee ID: {}".format(ID)
```

Let's create a new class called developer that inherits from Employee

```python
class Developer(Employee):
    pass
```

We can look at exactly what is inside of an class, and we can see that there are methods, class methods, static methods, and class variables all inherited from Employee

We literally didn't have to write any new code, and everything from Employee is just literally taken over

```python
print(help(Developer))
```

But... we still wanted Developers to have somethings that are unique to them over a general Employee

Like, maybe the developers would have a main programming language, and an experience level.

But we got a new problem again, how do I use the init method from the Employees Class, with the addition of a main programming language and an experience level for the Developer Class?

The reason that we do this is because we do not want to repeat code, and instead just want to build upon existing code.

Easy, we can just use `super()__init__` to initialize the attributes for an object from the inherited class, then initialize the attributes from the Developer Class

```python
class Developer(Employee):
    
    def __init__(self, first, last, pay, days_worked, main_prog_language, experience_lv):
        super().__init__(first, last, pay, days_worked)
        self.main_prog_language = main_prog_language
        self.experience_lv = experience_lv
```

See? Now we have developers! A type of customized employee

```python
dev_1 = Developer('Corey', 'Schafer', 50000, 5, "Python", "Advanced")
dev_2 = Developer('Test', 'Employee', 60000, 3, "Java", "Intermediate")

print(dev_1.email, ",", dev_1.main_prog_language)
```

While that is all cool and good, let's create another class for strengthening understanding, called managers, which of course manages a list of employees

```python
class Manager(Employee):
    
    def __init__(self, first, last, pay, days_worked, employees=None):
        super().__init__(first, last, pay, days_worked)
        if employees == None:
            self.employees = []
        else:
            self.employees = employees

    def add_employee(self, employee):
        if employee not in self.employees:
            self.employees.append(employee)

    def remove_employee(self, employee):
        if employee in self.employees:
            self.employees.remove(employee)

    def print_employees(self):
        for employee in self.employees:
            print(employee.fullname())
```

```python
mgr_1 = Manager('Sue', 'Smith', 90000, 5, [dev_1])
print(mgr_1.email)
```

```python
mgr_1.add_employee(dev_2)
mgr_1.print_employees()
```

```python
mgr_1.remove_employee(dev_2)
mgr_1.print_employees()
```

Let's also see our original functionalities working in place for Employee!

```python
print(mgr_1.eligible_for_break())
```

Though do be careful using too much inheritance... it get's all the classes, and subclasses jumbled up

When you need to change something in 1 class, it might be entangled to the functionalities of another and in changing one you have to change both, or even more!
