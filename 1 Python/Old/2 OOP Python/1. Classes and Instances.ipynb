# Why Classes?

It's a easy way to group functionalities of code together, and be used repeatedly

Let's get some terminology out of the way first:

- class: a blueprint/architecture for objects

- object: an instance of a class

- attribute: the properties of an object

- methods: functions that are associated with the class's object

A good question to ask is, why do we need "self" when we call methods on a object?

Well, "self" refers the instance of the object, so when we call "self" in a method we get access to all the attributes of that object, and can use it in our function

Also... python kinda forces you to do it, not having self... is another story for another day called class methods

```python
class Student: #a class

    def __init__(self, first_name, last_name, age, score): # initializing an object of a class
        self.first_name = first_name  # an attribute
        self.last_name = last_name
        self.age = age
        self.score = score

    def calculate_grade(self): # a method
        return 'A' if self.score >= 80 else 'B' if self.score >= 60 else 'C' if self.score >= 40 else 'F'
    
    def fullname(self):
        return f"{self.first_name} {self.last_name}"
    

#an object of class "Student"
student1 = Student("alice", "rosa", 15, 70)

#using a class method on the object "alice"
print(student1.fullname())
print(student1.calculate_grade())
```

If you directly try to print the object out, it will print out the location of the object in memory

```python
print(student1)
```

There are special "methods" that tell python what it means to "print" this object out, or basically what should python do when it needs to treat this object like a string
