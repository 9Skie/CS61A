# Class Methods

If you see previously... all the functions under our class had to pass in an argument, `self`, which just refers to the object

There are methods which don't take the object, but the class, they are called class methods

We will use the same class from before, but with the addition of a class method to see what it does

```python
class Employee:

    break_allowance_days = 4

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"
    
    def eligible_for_break(self):
        if self.days_worked > Employee.break_allowance_days:
            return True
        else:
            return False

    # our class method
    @classmethod
    def define_allowance_days(cls, days):
        if days < 7:
            cls.break_allowance_days = days 
    

emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)
```

So, you can essentially think it this way, under a class, we can create objects of a class, which have their attributes.

But the class itself can have these variables, sort of like attributes of the objects, but now attributes of a class.

So, our normal functions under a class acts on the objects of the class, a `classmethod` acts on the class itself instead

This class method, `define_allowance_days` acts on the class variable `break_allowance_days`

```python
# Before
print(Employee.break_allowance_days)
```

```python
Employee.define_allowance_days(days=5)

# After
print(Employee.break_allowance_days)
```

# Static Methods

These don't pass anything in! Not a class, not a object! They basically behave like a normal function, except for the fact that it's most likely linked to the functionality of our class in some way

So therefore, we still include it as a part of our class, but we need to say that it doesn't need to pass anything in, so basically say that it is a static method

Like in this case, we made a method which generates an ID for an Employee

```python
import random

class Employee:

    break_allowance_days = 4

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"
    
    def eligible_for_break(self):
        if self.days_worked > Employee.break_allowance_days:
            return True
        else:
            return False

    # our class method
    @classmethod
    def define_allowance_days(cls, days):
        if days < 7:
            cls.break_allowance_days = days 

    # our static method
    @staticmethod
    def generate_ID():
        ID = random.randint(10000000, 99999999)
        return "Employee ID: {}".format(ID)
```

```python
print(Employee.generate_ID())
```
