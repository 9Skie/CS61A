# Class Variables

Let's use this code for defining an Employee Class Instead

```python
class Employee:

    def __init__(self, first, last, pay):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay

    def fullname(self):
        return f"{self.first_name} {self.last_name}"
    

emp_1 = Employee('Corey', 'Schafer', 50000)
emp_2 = Employee('Test', 'Employee', 60000)
```

The idea of a class variable is that there might be common values that are shared among all instances of a class

So instead of giving each instance the same variable, we can use a class variable, which is the same amongst all instances of the class

Say... Every employee is allowed a break when they've worked for more than 4 days a week, so we can give a class variable "Break_Allowance_Days", and a function based on that to see if they can go on break or not

```python
class Employee:

    break_allowance_days = 4 # this is the class variable

    def __init__(self, first, last, pay, days_worked):
        self.first = first
        self.last = last
        self.email = first + '.' + last + '@email.com'
        self.pay = pay
        self.days_worked = days_worked

    def fullname(self):
        return f"{self.first} {self.last}"
    
    #this function uses the class variable
    def eligible_for_break(self):
        if self.days_worked > Employee.break_allowance_days:
            return True
        else:
            return False
    

emp_1 = Employee('Corey', 'Schafer', 50000, 5)
emp_2 = Employee('Test', 'Employee', 60000, 3)
```

Let's see if our first employee can have a break

```python
print(emp_1.eligible_for_break())
```

And since it's a class variable, we'd have to access it through the CLASS and not the OBJECTS of the class

```python
print(Employee.break_allowance_days)
```

While this works... the object itself doesn't have this attribute, so python checks if the class itself has this attribute, and returns that as the result

Which in this case, break_allowance_days doesn't exist for this object, but it does for the class, so python returns that

```python
print(emp_1.break_allowance_days)
```

BUT you can change this break_allowance_days for each object when we access it through the instance

```python
emp_1.break_allowance_days = 5
print(emp_1.break_allowance_days)
```
