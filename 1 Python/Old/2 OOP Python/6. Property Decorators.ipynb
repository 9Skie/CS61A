# Property Decorators

This... is a more complex topic, so let's get into it with a simplified employee class

```python
class Employee:

    def __init__(self, first, last):
        self.first = first
        self.last = last
        self.email = first + "." + last + "@email.com"

    def fullname(self):
        return f'{self.first} {self.last}'
```

When we have an employee instance created, these are the attributes of him

```python
emp_1 = Employee('John', 'Smith')
print(emp_1.first)
print(emp_1.email)
print(emp_1.fullname())
```

But... let's say we changed his first name, his other attribute "email", which is originally dependent on his first name and last name, isn't changed occordingly to how we like it

So how do we change that to act how we like it to? Well, we could just turn email into a method like full name

But that would break the existing structure of the code, and anyone else using this class would have to update their code accordingly, so we don't wanna do that

```python
emp_1.first = "Jim"
print(emp_1.first)
print(emp_1.email + " (This is not changed accordingly!)")
print(emp_1.fullname())
```

So we use something called a `property`, basically it makes email a method that acts like an attribute (getter = property, but this is just python syntax)

What the heck does that mean? Well, if we call it as a method, it would be `employee.email()`, but we can call a property as `employee.email`

Why do this? Because we can have <b>attributes that are defined used methods</b>

```python
class Employee:

    def __init__(self, first, last):
        self.first = first
        self.last = last

    @property
    def email(self):
        return f"{self.first}.{self.last}@email.com"

    @property
    def fullname(self):
        return f'{self.first} {self.last}'
```

Now let's try this again, everything works!

```python
emp_1 = Employee('John', 'Smith')
emp_1.first = "Jim"
print(emp_1.first)
print(emp_1.email)
print(emp_1.fullname)
```

Right now our email, and full name are dependent on the first, and last name, updating first or last name will change email and full name accordingly

But... we can't directly change the email and full name, that would raise and error, and we can't let updating these attributes reflect back to update the first and last name

So how do we do that?

```python
emp_1.fullname = "Jim Hoon"
```

We can use a setter in this case

`property` (getter) allows us to "get" this attribute using a method through other attributes

`setter` defines how we can directly "set" this attribute, not through other attributes

I mean... if you like to make the syntax concise, you can also use `@email.getter`, it's equivalent to `property`

```python
class Employee:

    def __init__(self, first, last):
        self.first = first
        self.last = last

    @property
    def email(self):
        return f"{self.first}.{self.last}@email.com"
    
    @email.setter
    def email(self, value):
        
        # assume it's a email
        if "@" in value:
            self.first, self.last = value.split("@")[0].split(".")
        # assume it's a full name
        else:
            self.first, self.last = value.split()
    
        return f"{self.first}.{self.last}@email.com"


    @property
    def fullname(self):
        return f'{self.first} {self.last}'
    
    @fullname.setter
    def fullname(self, fullname):
        self.first, self.last = fullname.split(" ")
        return f'{self.first} {self.last}'
```

```python
emp_1 = Employee('John', 'Smith')
emp_1.first = "Jim"
print(emp_1.first)
print(emp_1.email)
print(emp_1.fullname)
```

Now let's try updating the full name, and see how that reflects to the email, first, and last name

```python
emp_1.fullname = "Jim Hoon"
print(emp_1.first)
print(emp_1.last)
print(emp_1.email)
```

Now, getter, setter... the last is `deleter`

`deleter` defines what does it mean to directly "delete" this attribute, and how it should affect other related attributes 

```python
class Employee:

    def __init__(self, first, last):
        self.first = first
        self.last = last

    @property
    def email(self):
        return f"{self.first}.{self.last}@email.com"
    
    @email.setter
    def email(self, value):
        
        # assume it's a email
        if "@" in value:
            self.first, self.last = value.split("@")[0].split(".")
        # assume it's a full name
        else:
            self.first, self.last = value.split()
    
        return f"{self.first}.{self.last}@email.com"
    
    @email.deleter
    def email(self):
        print("Email Deleted")


    @property
    def fullname(self):
        return f'{self.first} {self.last}'
    
    @fullname.setter
    def fullname(self, fullname):
        self.first, self.last = fullname.split(" ")
        return f'{self.first} {self.last}'
    
    @fullname.deleter
    def fullname(self):
        print("Name Deleted")
        self.first = None
        self.last = None
        
```

Now let's delete the full name of our employee

```python
del(emp_1.fullname)
print(emp_1.first)
print(emp_1.last)
```
